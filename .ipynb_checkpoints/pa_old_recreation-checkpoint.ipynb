```python

inten_med = [
np.median(ps20nm15.peak_intens[ps20nm15.filter]**(1/6)),
np.median(ps30nm15.peak_intens[ps30nm15.filter]**(1/6)),
np.median(ps40nm15.peak_intens[ps40nm15.filter]**(1/6)),
np.median(ps50nm15.peak_intens[ps50nm15.filter]**(1/6)),
]

inten_err = [
np.std(ps20nm15.peak_intens[ps20nm15.filter]**(1/6)),
np.std(ps30nm15.peak_intens[ps30nm15.filter]**(1/6)),
np.std(ps40nm15.peak_intens[ps40nm15.filter]**(1/6)),
np.std(ps50nm15.peak_intens[ps50nm15.filter]**(1/6)),
]

ps_size = np.array([20, 30, 40, 50])

popt1, pcov1 = scipy.optimize.curve_fit(lambda x,m,c: m*x+c, ps_size, inten_med)
cov_m_m = pcov1[0,0]
cov_c_c = pcov1[1,1]
cov_m_c = pcov1[0,1]
# Error propagation
sigma_y = np.sqrt(
    (ps_size**2)*cov_m_m +
    cov_c_c +
    2*ps_size*cov_m_c
)

popt2, pcov2 = scipy.optimize.curve_fit(lambda x,m: m*x, ps_size, inten_med)

plt.figure()
x = np.linspace(10, 60)
plt.errorbar(x=ps_size, y=inten_med, yerr=inten_err,marker='.', ls='None')
plt.plot(x, popt1[0]*x+popt1[1])
#plt.plot(x, popt2*x)
```
